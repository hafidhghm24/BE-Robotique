```python
import numpy as np
```

```python
#CONSTANTES
pi2 = np.pi/2 #90°
pi = np.pi
```

```python
def MGD(q):
    q1,q2,q3,q4,q5,q6 = q[0],q[1],q[2],q[3],q[4],q[5]
    theta = [q1,q2,q3,q4,q5,q6]
    alpha = [float(0), -pi2, pi, float(0), pi, -pi2]
    r = [float(0.375), float(0), float(0.020), float(0.450), float(0), float(0.07)]
    a = [float(0),float(0),float(0.400),float(0),float(0),float(0)]
    transform = []
    for i in range(len(q)):
        T = np.array([
            [np.cos(theta[i]), -np.sin(theta[i]), 0.0, a[i]],
            [np.sin(theta[i])*np.cos(alpha[i]), np.cos(theta[i])*np.cos(alpha[i]), -np.sin(alpha[i]), -np.sin(alpha[i])*r[i]],
            [np.sin(theta[i])*np.sin(alpha[i]), np.cos(theta[i])*np.sin(alpha[i]),  np.cos(alpha[i]),  np.cos(alpha[i])*r[i]],
            [0.0, 0.0, 0.0, 1.0]
        ])
        transform.append(T)
    return transform
    
```

```python
#configuration
q=[0,pi2,0,-pi2,pi2,-pi2]

print(f"configuration: {q}\n")

#affichage transformations 
T = MGD(q)

#print(f"transforme: {T}")
for i in range(len(T)):
    print(f"transforme[{i}_{i+1}]:")
    print(T[i])
    print()
```
