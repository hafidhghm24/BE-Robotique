```python
import numpy as np
```

```python
#CONSTANTES
pi2 = np.pi/2 #90° en radians 
pi = np.pi
```

```python
def MGD(q):
    '''retourne la matrice de chaque Ti,i+1 AINSI que la matrice de transformation globale T0,6'''
    q1,q2,q3,q4,q5,q6 = q[0],q[1],q[2],q[3],q[4],q[5]
    theta = np.array([q1,q2,q3,q4,q5,q6])
    alpha = np.array([0, -pi2, pi2, 0, pi2, -pi2])
    r = np.array([0.375, 0, 0.02, 0.45, 0, 0.07])
    a = np.array([0, 0, 0.4, 0, 0, 0])
    transform = []
    T06 = np.eye(4)
    for i in range(len(q)):
        T = np.array([
            [np.cos(theta[i]), -np.sin(theta[i]), 0.0, a[i]],
            [np.sin(theta[i])*np.cos(alpha[i]), np.cos(theta[i])*np.cos(alpha[i]), -np.sin(alpha[i]), -np.sin(alpha[i])*r[i]],
            [np.sin(theta[i])*np.sin(alpha[i]), np.cos(theta[i])*np.sin(alpha[i]),  np.cos(alpha[i]),  np.cos(alpha[i])*r[i]],
            [0, 0, 0, 1]
        ])
        transform.append(T)
        T06 = np.dot(T06, T)

    return transform, T06
```

```python
#configuration
qfig=[0,pi2,0,-pi2,pi2,-pi2]

print(f"configuration: {qfig}\n")

#affichage transformations 
T, T06 = MGD(qfig)

#print(f"transforme: {T}")
# for i in range(len(T)):
#     print(f"transforme[{i}_{i+1}]:")
#     print(T[i])
#     print()
print(f"Matrice de transformation T0,6 :\n {np.round(T06,2)}")
```
